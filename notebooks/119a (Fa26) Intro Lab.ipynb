In our course, we'll refer to "parameters" as a fixed value that we refer back to by a particular name.  Unlike input and output variables, parameters will not change values, although we may reference a parameter in computations later on.

Telling Sage to remember a particular parameter value or a function will make our future computations shorter because we won't have to write out all the individual steps each time; we can just ask Sage to bring back the values that we told it to remember for us. If we include an equals sign, we are telling Sage to remember something.  If we do not include an equals sign, then we are asking Sage to actually compute something.  In the parts below, pay attention to when we want to ask Sage to remember something for us versus when we want it to use the values it is remembering to actually compute something.
